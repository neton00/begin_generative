# AIGC란?
- ai generated content의 약자로, 인공지능 기술을 활용하여 텍스트, 오디오, 이미지, 비디오등 다양한 형태의 콘텐츠를 생성하는 것을 말한다.

- AIGC 기술의 4가지 범주로는 텍스트, 오디오, 이미지 및 비디오, 가상 공간이 있다

## 텍스트 범주
- 글쓰기 로봇, 챗봇과 같이 텍스트 콘텐츠를 생성하거나 편집하는 것

## 오디오 범주
- 텍스트-오디오 변환, 음성 변환 등을 포함하여 오디오 콘텐츠를 생성하거나 편집하는 것

## 이미지 및 비디오 범주
- 얼굴 생성, 이미지 편집, 동영상 콘텐츠 생성 등

## 가상공간 범주
- 3D 모델링, 디지털 트윈, 가상 장면 편집 등

## AIGC 애플리케이션 관점에서 AIGC의 큰 이점은?
- 풍부하고 다양하며 역동적이고 상호작용이 풍부한 콘텐츠를 제공할 수 있음.
- 정확성, 저작권, 비용 문제가 사라지는 것은 아님

## diffusion model의 정의
- 확산 모델은 고차원의 복잡한 데이터의 확률 분포를 모델링하기 위한 생성 모델의 한 종류다.
- 확산 모델은 확산과정을 기반으로 데이터의 생성 과정을 설명하고, 역확산 과정을 통해 사후 확률에서 사전 확률 분포를 점진적으로 추론하여 고차원 복잡 데이터의 모델링을 달성하는 것이다.

## diffusion model의 개발과정
- 랑주뱅 역학(langevin dynamics) -> 노이즈 제거 점수 매칭 -> 확산 과정 -> 비모수적 확산 프로세스 -> 확산 모델

## 랑주뱅 역학
- score를 이용해서 확률분포를 따라가며 샘플을 생성하는 기본적인 샘플링방법
- 입자를 에너지에 따라 움직이게 하면서 동시에 랜덤한 노이즈도 섞어서 결국 원하는 확률 분포에서 샘플을 뽑는 방법
- 핵심 : 학습한 score를 이용해 실제 샘플을 생성

## 노이즈 제거 점수매칭(Denoising Score Matching, DSM)
- 깨끗한 데이터에 노이즈를 넣고 신경망이 이 노이즈를 제거하려면 어느 방향으로 움직여하 하는가를 배우게 하는 방법
- s(x)=∇x​logp(x) -> score function : 현재 x에서 데이터가 더 많이 존재하는 방향을 알려주는 벡터
- 직관적으로 score = noisy sample을 원래 데이터 쪽으로 돌려 놓는 방향
- 핵심 : 노이즈를 섞은 데이터를 이용해서 score를 학습

## score matching
- 데이터 분포의 score를 학습

## Gaussian noise
- 평균 0을 중심으로, 정규분포(Gaussian distribution)를 따르는 랜덤한 값들을 데이터에 더하는 것

## Duffusin process
- 쉽게 말하면 어떤 데이터에 시간이 지나면서 점점 랜덤한 노이즈가 섞여 가능 과정을 수학적으로 표현한 것

- 확산 과정에서 가장 기본적인 형태 중 하나가 확률미분방정식(SDE)
- dx=f(x,t)dt+g(t)dW
	​f(x,t)dt -> 결정론적인 이동
    g(t)dW -> 랜덤한 움직임
    dW -> brownian motion(브라운 운동) / Wiener process(위너 과정)의 랜덤한 변화
- 확산 과정에서는 이 랜덤성을 아주 작은 시간 단위에 계속 넣는다 생각하면 됨

## Brownian motion
- 예측할 수 없는 방향으로 계속 움직이는 무작위 움직임

## Wiener process
- 브라운 운동을 수학적으로 표현한 대표적인 모델 Wiener process Wt
- 아주 짧은 시간 동안의 변화량은 평균 0인 가우시안 랜덤값이다.

가우시안 노이즈 -> 한번의 랜덤한 변화
Wiener process -> 그 랜덤한 변화를 시간에 따라 계속 누적한것
Diffusin process -> Wiener proecess를 이용하여 데이터가 시간에 따라 변화하도록 만든 확률 과정

## Non-parametric Diffusion Process(비모수적 확산 프로세스)
- 데이터 분포 p(x)의 모양을 직접 가정하지 않고, 데이터에서 score를 학습해서 확산 과정을 다루는 것.
- 몇개의 파라미터(parameter)로 데이터 분포를 표현하는 것이 모수적 접근이다(parametric)
- 실제 데이터 분포는 굉장히 복잡한 고차원 분포이므로 "이 데이터는 Gaussian이다"처럼 특정한 분포형태를 미리 가정하기 어려움 -> 그래서 비 모수적 접근을 사용

## non-parametric
- 데이터 분포의 형태를 미리 특정한 몇 개의 파라미터로 정해놓지 않음
- 그렇다 해서 아무것도 학습하지 않는것이 아니라 신경망을 이용해서 데이터 분포의 중요한 정보를 학습함
- 그게 Score. 로 Score야. s(x)=∇xlogp(x)

## 왜 분포 자체가 아니라 Score를 배우는지?
- 우리가 원하는건 p(x)자체가 아니라 그 분포에서 샘플을 생성하는 것임
- 랑주뱅 역학에서도 필요한건 p(x)가 아니라 dx=∇x​logp(x)dt+루트2dW 에서 ∇xlogp(x)임
- 즉 분포의 정확한 모양을 알 필요없이, 각 위치에서 어느 방향으로 가야 데이터가 있는지 알면 샘플링할 수 있다.

##  Foward dffusion


- 이전 단계 x_t-1가 주어졌을 대, 다음 단계 x_t fmf 어떻게 만들것인지
- 베타는 t번째 단계에서 얼마나 많은 노이즈를 추가할지 결정한는 겂
- 이 프로세스에서 학습은 없고 노이즈를 정해주면 그냥 계산으로 결정됨

## Reverse process

- pθ​(xt−1​∣xt​)
- 현재 noisy image x_t가 주어졌을 때, 한 단계 전인  x_t-1은 어떤 분포일까?
- 세타는 신경망의 파라미터


| $q(\mathbf{x}_t\vert\mathbf{x}_0)=\mathcal N(\mathbf{x}_t;\sqrt{\bar\alpha_t}\mathbf{x}_0,(1-\bar\alpha_t)\mathbf I)$ | 문제 2 |
- x_0에서 원하는 x_t를 한 번에 만들 수 있다
- 가우시안의 합성 성질 덕분에 닫힌 형식이 되어 T번 반복하지 않고 한 번에 시점 t로 확신시킬 수 있다.
